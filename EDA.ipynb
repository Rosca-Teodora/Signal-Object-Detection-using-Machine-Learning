```python
import pandas as pd
import os

# load labels
data_dir = 'data/'
train_csv_path = os.path.join(data_dir, 'train.csv')
df = pd.read_csv(train_csv_path)

df['filename'] = df['id'].astype(str) # copy id into dataframe

print(df.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))
ax = sns.countplot(data=df, x='label', hue='label', palette='viridis', legend=False)
plt.title('Class Distribution (Object Counts 1 to 5)')
plt.xlabel('Number of Objects (Label)')
plt.ylabel('Number of Images')

# observe object count in case there will be a need for object weighing later on
for p in ax.patches:
    ax.annotate(f'{int(p.get_height())}', 
                (p.get_x() + p.get_width() / 2., p.get_height()), 
                ha='center', va='bottom', fontsize=10)

plt.show()
```
